# Exploratory Data Analysis — Iris Dataset

**Objective:** Analyze the prepared Iris dataset with descriptive statistics and visualizations, then report at least five insights relevant to modelling or decisions.

**Dataset file:** `iris_cleaned.csv` (keep it in the same folder as this notebook).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv("iris_cleaned.csv")
display(df.head())
```

## 1. Dataset structure and data quality

```python
print("Shape (rows, columns):", df.shape)
print("\nColumn data types:")
print(df.dtypes)
print("\nMissing values per column:")
print(df.isna().sum())
print("\nExact duplicate rows:", df.duplicated().sum())
print("\nSpecies counts:")
print(df["species"].value_counts())
```

## 2. Descriptive statistics

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
display(df[numeric_cols].describe().T)
print("Mean measurements by species:")
display(df.groupby("species")[numeric_cols].mean().round(2))
print("Median measurements by species:")
display(df.groupby("species")[numeric_cols].median().round(2))
```

## 3. Visualizations

```python
# A. Histograms
df[numeric_cols].hist(figsize=(12, 8), bins=15, edgecolor="black")
plt.suptitle("Distributions of Numerical Features")
plt.tight_layout()
plt.show()

# B. Species balance
sns.countplot(data=df, x="species")
plt.title("Species Distribution")
plt.xlabel("Species")
plt.ylabel("Sample Count")
plt.show()

# C. Correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# D. Box plots by species
melted = df.melt(id_vars="species", value_vars=numeric_cols, var_name="feature", value_name="measurement_cm")
plt.figure(figsize=(10, 6))
sns.boxplot(data=melted, x="feature", y="measurement_cm", hue="species")
plt.title("Feature Distributions by Species")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

# E. Petal relationship
sns.scatterplot(data=df, x="petal_length_cm", y="petal_width_cm", hue="species", s=60)
plt.title("Petal Length vs Petal Width by Species")
plt.tight_layout()
plt.show()
```

## 4. Insights (based on this dataset)

```python
# Calculate correlations and species-level means used in the written findings
corr = df[numeric_cols].corr()
print("Petal length vs petal width correlation:", round(corr.loc["petal_length_cm", "petal_width_cm"], 3))
print("Sepal length vs petal length correlation:", round(corr.loc["sepal_length_cm", "petal_length_cm"], 3))
print("Sepal width vs sepal length correlation:", round(corr.loc["sepal_width_cm", "sepal_length_cm"], 3))
display(df.groupby("species")[["petal_length_cm", "petal_width_cm"]].mean().round(2))

print("
Rows labelled setosa with petal width above 0.6 cm:")
display(df[(df["species"] == "Iris-setosa") & (df["petal_width_cm"] > 0.6)])
print("Exact duplicate rows:")
display(df[df.duplicated(keep=False)])
```

### Interpretation and recommended actions

1. **Balanced classes:** each species has 50 records. Preserve this balance with stratified train/test splitting.
2. **Petal features are promising predictors:** mean petal length and width differ markedly by species; test these features in a classifier.
3. **Petal features are strongly correlated:** they contain overlapping information; compare models with both features and a simpler feature set.
4. **Sepal length is associated with petal measurements:** it may also help classification, but confirm using validation.
5. **Sepal width is less strongly correlated with sepal length:** it may offer complementary information; assess empirically.
6. **Suspicious setosa measurement:** inspect the row with petal width 1.3 cm against the source before editing it.
7. **One exact duplicate:** verify whether it is accidental; identical records can inflate evaluation if they leak across train/test sets.
8. **Potential outliers:** use box plots as screening tools only; do not remove observations without investigation.

## Conclusion
The dataset is balanced and shows strong species-related patterns in petal measurements. Check the suspicious measurement and exact duplicate before modelling. Use stratified validation and report per-class precision, recall, F1-score, and a confusion matrix as well as accuracy.
